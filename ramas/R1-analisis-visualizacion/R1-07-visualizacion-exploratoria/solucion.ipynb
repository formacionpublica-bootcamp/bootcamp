```python
# ── Preparación del entorno (ejecuta esta celda primero) ──────────────────────
import os, urllib.request
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

# "En vivo o caché": usa el archivo local; si falta (ej. Colab), lo baja del repo.
CSV = "matricula.csv"
if not os.path.exists(CSV):
    urllib.request.urlretrieve(f"https://raw.githubusercontent.com/formacionpublica-bootcamp/bootcamp/main/data/matricula.csv", CSV)

df = pd.read_csv(CSV)
print("Datos cargados:", df.shape, "filas x columnas")
df.head()
```

### Datos cargados arriba. Exploremos viendo.

## 1. Histograma: la forma de una variable

El **histograma** muestra cómo se distribuye un número. El tamaño de los establecimientos tiene cola larga: muchas escuelas chicas y pocos colegios enormes (recortamos al p99 para verlo mejor).

### ✍️ Ejercicio 1 — Dibuja el histograma de la matrícula

```python
matriculas = df["mat_total"].clip(upper=df["mat_total"].quantile(0.99))
fig, ax = plt.subplots(figsize=(6, 4))
ax.hist(matriculas, bins=40, color="#4240e5")
ax.set_xlabel("Matrícula (recortada al p99)"); ax.set_ylabel("Frecuencia"); ax.set_title("Tamaño de los establecimientos")
plt.show()
```

```python
# ── Celda de chequeo · Ejercicio 1 ──────────────────────────────────────────
try:
    assert fig.__class__.__name__ == "Figure"
    assert len(ax.patches) > 0
    print("✅ Ejercicio 1: ¡correcto!")
except Exception as e:
    print("❌ Aún no. Revisa tu respuesta y vuelve a intentarlo.")
    print("   Pista:", 'ax.hist(matriculas, bins=40).')
    print("   Detalle:", e)
```

## 2. Boxplot: comparar grupos

El **boxplot** compara la distribución entre grupos. ¿Difiere el tamaño de los establecimientos según su dependencia?

### ✍️ Ejercicio 2 — Boxplot de la matrícula por dependencia

```python
orden = ["Municipal", "Particular subvencionado", "Particular pagado", "Servicio Local de Educación"]
tope = df["mat_total"].quantile(0.99)
grupos = [df[df["dependencia"] == d]["mat_total"].clip(upper=tope) for d in orden]
fig, ax = plt.subplots(figsize=(8, 4))
ax.boxplot(grupos, labels=["Municipal", "P. subv.", "P. pagado", "SLEP"])
ax.set_ylabel("Matrícula"); ax.set_title("Matrícula por dependencia")
plt.show()
```

```python
# ── Celda de chequeo · Ejercicio 2 ──────────────────────────────────────────
try:
    assert fig.__class__.__name__ == "Figure"
    assert len(grupos) == 4
    print("✅ Ejercicio 2: ¡correcto!")
except Exception as e:
    print("❌ Aún no. Revisa tu respuesta y vuelve a intentarlo.")
    print("   Pista:", 'ax.boxplot(grupos, labels=[...]).')
    print("   Detalle:", e)
```

## 3. Dispersión: ¿se relacionan dos variables?

El **gráfico de dispersión** revela relaciones. ¿Más cursos implica más estudiantes? ¿Y qué tan dispersa es esa relación?

### ✍️ Ejercicio 3 — Dispersión cursos vs matrícula

```python
m = df.sample(800, random_state=42)
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(m["cursos"], m["mat_total"], s=10, alpha=0.4, color="#ff6b35")
ax.set_xlabel("Nº de cursos"); ax.set_ylabel("Matrícula"); ax.set_title("Cursos vs matrícula")
plt.show()
```

```python
# ── Celda de chequeo · Ejercicio 3 ──────────────────────────────────────────
try:
    assert fig.__class__.__name__ == "Figure"
    assert len(ax.collections) > 0
    print("✅ Ejercicio 3: ¡correcto!")
except Exception as e:
    print("❌ Aún no. Revisa tu respuesta y vuelve a intentarlo.")
    print("   Pista:", "ax.scatter(m['cursos'], m['mat_total']).")
    print("   Detalle:", e)
```

## 4. Barras: comparar categorías

Las **barras** comparan magnitudes entre categorías. ¿Qué región concentra más estudiantes?

### ✍️ Ejercicio 4 — Matrícula por región (barras horizontales)

```python
g = df.groupby("region")["mat_total"].sum().sort_values()
fig, ax = plt.subplots(figsize=(6, 5))
ax.barh(g.index, g.values, color="#4240e5")
ax.set_xlabel("Estudiantes matriculados"); ax.set_title("Matrícula por región")
plt.show()
```

```python
# ── Celda de chequeo · Ejercicio 4 ──────────────────────────────────────────
try:
    assert fig.__class__.__name__ == "Figure"
    assert len(g) == df["region"].nunique()
    print("✅ Ejercicio 4: ¡correcto!")
except Exception as e:
    print("❌ Aún no. Revisa tu respuesta y vuelve a intentarlo.")
    print("   Pista:", 'ax.barh(g.index, g.values).')
    print("   Detalle:", e)
```
