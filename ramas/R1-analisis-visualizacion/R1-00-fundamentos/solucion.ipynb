```python
# ── Preparación del entorno (ejecuta esta celda primero) ──────────────────────
import os, urllib.request
import pandas as pd
import matplotlib.pyplot as plt

# "En vivo o caché": usa el archivo local; si falta (ej. Colab), lo baja del repo.
CSV = "matricula.csv"
if not os.path.exists(CSV):
    urllib.request.urlretrieve(f"https://raw.githubusercontent.com/formacionpublica-bootcamp/bootcamp/main/data/matricula.csv", CSV)

df = pd.read_csv(CSV)
print("Datos cargados:", df.shape, "filas x columnas")
df.head()
```

### Datos cargados arriba. Empecemos por lo básico de Python.

## 1. Variables: guardar para reutilizar

Una **variable** guarda un valor con nombre. Cada fila del dataset es un establecimiento educacional de Chile con su matrícula 2025. Tomemos la matrícula total de cada uno como una lista y calculemos cuántos estudiantes hay en total y el promedio por establecimiento, sin pandas.

### ✍️ Ejercicio 1 — Total de estudiantes y promedio por establecimiento

```python
matriculas = df["mat_total"].tolist()
total = sum(matriculas)
promedio = total / len(matriculas)
print("estudiantes en Chile:", total, "| promedio por establecimiento:", round(promedio))
```

```python
# ── Celda de chequeo · Ejercicio 1 ──────────────────────────────────────────
try:
    assert total == int(df["mat_total"].sum())
    assert abs(promedio - df["mat_total"].mean()) < 1e-6
    print("✅ Ejercicio 1: ¡correcto!")
except Exception as e:
    print("❌ Aún no. Revisa tu respuesta y vuelve a intentarlo.")
    print("   Pista:", 'sum(matriculas) y total / len(matriculas).')
    print("   Detalle:", e)
```

## 2. Listas y comprensiones: filtrar

Una **comprensión de lista** filtra en una línea. Contemos los establecimientos grandes.

### ✍️ Ejercicio 2 — Cuenta los establecimientos con más de 1.000 estudiantes

```python
grandes = [m for m in matriculas if m > 1_000]
n_grandes = len(grandes)
print("establecimientos con más de 1.000 estudiantes:", n_grandes)
```

```python
# ── Celda de chequeo · Ejercicio 2 ──────────────────────────────────────────
try:
    assert n_grandes == int((df["mat_total"] > 1_000).sum())
    print("✅ Ejercicio 2: ¡correcto!")
except Exception as e:
    print("❌ Aún no. Revisa tu respuesta y vuelve a intentarlo.")
    print("   Pista:", '[m for m in matriculas if m > 1_000] y luego len(...).')
    print("   Detalle:", e)
```

## 3. Diccionarios y bucles: acumular por clave

Un **diccionario** asocia clave→valor. Con un **bucle** sumamos los estudiantes por región.

### ✍️ Ejercicio 3 — Estudiantes por región con un bucle

```python
est_region = {}
for region, m in zip(df["region"], df["mat_total"]):
    est_region[region] = est_region.get(region, 0) + m
print(len(est_region), "regiones")
```

```python
# ── Celda de chequeo · Ejercicio 3 ──────────────────────────────────────────
try:
    assert len(est_region) == df["region"].nunique()
    assert sum(est_region.values()) == int(df["mat_total"].sum())
    print("✅ Ejercicio 3: ¡correcto!")
except Exception as e:
    print("❌ Aún no. Revisa tu respuesta y vuelve a intentarlo.")
    print("   Pista:", 'est_region[region] = est_region.get(region, 0) + m.')
    print("   Detalle:", e)
```

## 4. Condicionales: clasificar en tramos

Con **if/elif/else** dentro de un bucle clasificamos cada establecimiento por tamaño.

### ✍️ Ejercicio 4 — Cuenta los establecimientos por tramo de tamaño

```python
tramos = {"chico": 0, "medio": 0, "grande": 0}
for m in matriculas:
    if m < 100:
        tramos["chico"] += 1
    elif m < 500:
        tramos["medio"] += 1
    else:
        tramos["grande"] += 1
print(tramos)
```

```python
# ── Celda de chequeo · Ejercicio 4 ──────────────────────────────────────────
try:
    assert sum(tramos.values()) == len(matriculas)
    assert tramos["chico"] == int((df["mat_total"] < 100).sum())
    print("✅ Ejercicio 4: ¡correcto!")
except Exception as e:
    print("❌ Aún no. Revisa tu respuesta y vuelve a intentarlo.")
    print("   Pista:", 'if m < 100 / elif m < 500 / else, sumando 1 al tramo correspondiente.')
    print("   Detalle:", e)
```

```python
# (ilustración) Establecimientos por tramo de tamaño
plt.figure(figsize=(5, 4))
plt.bar(tramos.keys(), tramos.values(), color="#4240e5")
plt.ylabel("Nº de establecimientos"); plt.title("Establecimientos por tamaño")
plt.tight_layout(); plt.show()
```
